```python
monthly = df.groupby('Month') # we are grouping by month
```

```python
monthly_precip = monthly['Daily_Precip_Total_mm'].mean()
monthly_precip
```

```python
july_temp = 11.2
july_precip = 2.67
print(f"At Tollick in july it was {july_temp} degrees celcius and it rained on average  {july_precip} mm!")
```

```python
december_temp = -21.44
december_precip = 0.4252
print(f"In the month of December at Tollick, it was {december_temp} degrees celcius and rained on average {december_precip} mmm!")
```

```python
yearly = df.groupby('Year')
yearly_means = yearly['Daily_AirTemp_Mean_C'].mean()
print(yearly_means)
```

```python
year_1988 = -4.88
year_1989 = -7.94
yearly_difference = year_1989 - year_1988
print(f"In the year 1988 it was {year_1988} degrees celcius and in the year 1989 it was {year_1989} degrees celcius. The temperature got colder on average {yearly_difference}.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://eds-217-essential-python.github.io/data/toolik_weather.csv"
df = pd.read_csv(url)

monthly = df.groupby('Month')
monthly_means = monthly['Daily_AirTemp_Mean_C'].mean()
```

```python
plt.plot(monthly_means)
plt.title("toolik monthly temperatures")
plt.xlabel("month")
plt.ylabel("temperature (C*)")
plt.show()
```

```python
plt.bar(monthly_means.index, monthly_means)
plt.title("average monthly temperatures")
plt.xlabel("month")
plt.ylabel("temperature (C*)")
plt.show()
```

```python
print(monthly_means)
```

```python
coldest = -22.9 # january's value, read from the chart
warmest = 11.22
print(f"tooliks coldest month averages about {coldest} degreees celcius (°C). However, the warmest month averages about {warmest} ")
```

```python
monthly_means.to_csv("monthly_means_temp.csv", header=True, float_format="%.2f")
```
