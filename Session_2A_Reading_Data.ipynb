```python
print("rows: ", goleta.shape[0])
print("columns: ", goleta.shape[1])

print(f"(rows, columns) = {goleta.shape}")
```

```python
goleta.columns.tolist() # shows the dataframes columns and puts it into a list. does not do too much
```

```python
goleta.info() # very important because it combines the shape, the column names, the types, and, critically, how many non-null 
# values each column has
```

```python
goleta['country_iso'] # viewing the column and its contents
```
