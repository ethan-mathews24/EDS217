### 2. The comparison operators

```python
print(temperature > 0) # greater then
print(temperature < 0) # less then
print(temperature >= 0) # greater or equal 
print(temperature <= 0) # less or equal
print(temperature == 8.4) # is equal to
print(temperature != 0) # not equal to


# watch out!
# temperature = 0 # this will cause a problem
```

```python
# string operator work but they are EXACT; no wiggle room

toolik.loc[0, 'Station'] == 'TLKMAIN'
```

### 3. Asking two questions at once

```python
temperature = 8.4
precipitation = 0.0

print(temperature > 0 and precipitation == 0.0) # and is true when both sides are true
print(temperature > 0 or precipitation > 0.0) # or is true when once side is true
print(not temperature > 0) # is not flips the answer


# what about &? 
```

```python
temperature = -12.0

if temperature > 10:
    category = 'mild'
elif temperature > 0:
    category = 'cool'
else:
    category = 'cold'

print(category)
```

### 5. Asking a whole column

```python
for value in toolik.loc[[0, 5000, 2], 'Daily_AirTemp_Mean_C']:
    if value > 7:
        print(f"{value} is warm")
    elif value > 0:
        print(f"{value} is mild")
    else:
        print(f"{value} is brrrr")

```

```python
# lets do a whole column at once

# Rules for logical operators and data frames/series
# 1 wrap every conditional statment in parantheses ()
# 2 when doing "and" and "or"s, we swtich to "&" and "|"s

warm_and_dry = (toolik['Daily_AirTemp_Mean_C'] > 0) & (toolik['Daily_Precip_Total_mm'] == 0)
print(warm_and_dry)
```

```python
warm_and_dryer = (toolik['Daily_AirTemp_Mean_C'] > 0) | (toolik['Daily_Precip_Total_mm'] == 0)
print(warm_and_dryer)
```

### 6. Counting the Trues

```python
above_zero = toolik['Daily_AirTemp_Mean_C'] > 0

print(above_zero.sum() / toolik.shape[0])

# either work

print(above_zero.sum() / len(toolik))

# also

print(sum(above_zero) / len(toolik))
```
