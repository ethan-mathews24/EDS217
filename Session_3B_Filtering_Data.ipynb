```python
import pandas as pd

url = 'https://eds-217-essential-python.github.io/data/openaq_goleta_measurments.csv'
goleta = pd.read_csv(url)
goleta.shape
```

```python
goleta.info()
```

```python
keep = goleta['value'] > 20 # can also assign this to a new object to save them

gt_20 = goleta[keep] # can also save this as a new object

gt_20.describe()

gt_20b = goleta[goleta['value'] > 20]

# Filter pattern is;
#
# df
#
# df_filtered = df[df['column'] > x]
# 
# can also wrap them 
#
# df_filtered = df[ (df['column'] > x) & (df['another column'] == y) ]
```

```python
valid = (goleta['value'] > 20)
high_readings = goleta[valid]

high_readings.shape
```

```python
goleta[valid].describe()
```

```python
# Filter with text 

goleta['parameter'].value_counts()
ozone = goleta[ goleta['parameter'] == 'o3']
ozone.head()
```

```python
ozone.shape
```

```python
goleta[goleta['parameter'] == 'O3'].shape
# An empty result is not an error. Zero rows is a real answer, 
# and it usually means a spelling or capitalization problem rather than a fact about the data.
```

```python
print(goleta['parameter'].unique())
pm25 = goleta[goleta['parameter'] == 'pm25']
```

```python
# combining logical operators

valid = (goleta['parameter'] == 'pm25') & (goleta['value'] > 9)
goleta[valid].describe()

```

```python
# can also put it all together
goleta_filtered = goleta[
    (goleta['parameter'] == 'pm25') & (goleta['value'] > 9)]


goleta_filtered
```

```python
# negation in pandas world (not not, but is ~)

goleta[~(goleta['parameter'] == 'pm25')]
```

```python
either_pollutant_high = goleta[
    ((goleta['parameter'] == 'pm25') & (goleta['value'] > 9))
    | ((goleta['parameter'] == 'pm10') & (goleta['value'] > 20))
]
either_pollutant_high.shape
```
