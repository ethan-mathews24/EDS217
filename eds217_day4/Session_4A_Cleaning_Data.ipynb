```python
plastics[['OBJECTID', 'GlobalID', 'Latitude', 'Longitude', 'Date', 'Measurement']].head(3)
```

```python
sample_column = ['Latitude', 'Longitude', 'Date', 'Measurement', 'Unit']

plastics.duplicated(subset=sample_column).sum()
```

```python
deduplicated = plastics.drop_duplicates(subset = sample_column)
deduplicated.shape
```

```python
repeats = plastics[plastics.duplicated(subset = sample_column, keep = False)]
print(repeats.shape)

repeats[['Measurement', 'Unit', 'Sampling Method', 'Short Reference']].head(6)
```

```python
# this is being applied to the whole dataset. typically when a line of code appplies
# to the whole DF then it is being used as a filter.
plastics.duplicated(subset = sample_column, keep = False)
```
