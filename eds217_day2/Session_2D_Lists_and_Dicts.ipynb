### 5. Making dictionaries

```python
names_pets = {'Nina': 1, 'Priscilla': 7, 'Elyse': 5, 'Karolina': 0}
```

```python
names_pets['Karla'] = 1
```

```python
names_pets
```

### 6. Getting things out of a dictionary

```python
names_pets['Nina'] # we are using the key to find how many pets nina has
```

### 7. Looping over a dictionary

```python
for key, value in names_pets.items(): 
    print(f"{key} has {value} pets")
    if value > 1:
        print(f"{key} has too many pets!")
```
