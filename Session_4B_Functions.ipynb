```python
# a function that can be used later to convert C° to F°
def celcius_to_fahrenheit(celcius):
    fahrenheit = (celcius * 9 / 5) + 32
    return fahrenheit
```

```python
celcius_to_fahrenheit(25)
```

```python
def ppm_to_ppb(concentration_ppm):
    """Convert a concentration from parts per million to parts per billion."""
    return concentration_ppm * 1000

# The line in triple quotes is a docstring. It is optional, it goes on the first line of the body, 
# and Python will show it back to you when you ask for help: `help(ppm_to_ppb)``
```

```python
def convert_depth(depth, units='meters'):
    """Convert a depth in metres to metres, feet, or fathoms."""
    if units == 'meters':
        return depth
    elif units == 'feet':
        return depth * 3.28084
    elif units == 'fathoms':
        return depth * 0.546807
    else:
        return None
```

```python

convert_depth(10, units = 'feet')
```

```python
convert_depth(10, 'feet')
# same as above
```

```python
def classify_latitude(latitude):
    if -23.5 <= latitude <= 23.5:
        return 'tropical'
    elif latitude < -66.5 or latitude > 66.5:
        return 'polar'
    else:
        return 'temperate'

```

```python
# pandas is taking the function and applying it to the entire column in 'Latitude'
plastics['Latitude'].apply(classify_latitude)

```

```python
# # 2. Apply it and assign it to a brand new column name
plastics['zone'] = plastics['Latitude'].apply(classify_latitude)
plastics.head()
```
