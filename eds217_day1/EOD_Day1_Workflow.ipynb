```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://eds-217-essential-python.github.io/data/toolik_weather.csv"
df = pd.read_csv(url)
```

here we are reading in the pandas as pd and pyplot as plt for future calls. The data set is being saved as the "url" object and then being read in from the panda function `pd.read_csv()`

```python
df.head()
```

here we are getting a preview of the dataframe to see the structure and what populates the rows.

```python
df.isnull().sum()
```

this function shows us the null values for each column. the .sum is a functioned added onto the df.null() function.

```python
monthly = df.groupby('Month')
monthly_means = monthly['Daily_AirTemp_Mean_C'].mean()
monthly_means
```

```python
plt.plot(monthly_means)
plt.title("Avarage Monthly Temperatures")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
monthly_means.to_csv("monthly_means.csv", header=True, float_format="%.2f")
```

```python
windspeed_avg = monthly['Daily_windsp_mean_msec'].mean()
windspeed_avg
```

```python
by_year = df.groupby('Year')
yearly_means = by_year['Daily_AirTemp_Mean_C'].mean()
yearly_means
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
plt.bar(months, monthly_means)
plt.title("Monthly Mean Air Temperature, 1988 to 2018")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.show()
```
